```python
import matplotlib as mpl
import matplotlib.pyplot as plt
import matplotlib.colors as colors
import networkx as nx
import csv
import json
import geopandas as gpd
import numpy as np
import math

with open(f"Graph_json/Graph_mean_farm_and_coast_PD60.json", "r") as f:
    data = json.load(f)
G = nx.node_link_graph(data)

farm_file = '/projappl/project_2018610/data/DTO4OWE_polygons/Wind_farms_Bothnian_Sea.shp'
coast_file = '/projappl/project_2018610/data/coastal/nemo_nordic_10-20m_polygons.shp'
gdf_farm = gpd.read_file(farm_file)
gdf_coast = gpd.read_file(coast_file)

farmIDs = np.arange(1, 20)
coastIDs = gdf_coast.sort_values('fid')['ID'].astype(str).tolist() # creating a list of coastal areas sorted by fid
nb_f = 19 # number of OWF
nb_c = 26 # number of coastal areas
```

```python
H = G.copy()
for nd in list(H.nodes()):
    if (nd, nd) in list(H.edges()):       
        H.remove_edge(nd, nd)
for e in list(H.edges()):
    H.edges[e[0], e[1]]['weight'] = - math.log(G.edges[e[0], e[1]]['prob'])
    # the path is shorter when the probability is higher, using -log because we need to multiply probabilities
```

### Calculating shortest paths between coastal areas ###

```python
stepping_stones_source = {}
stepping_stones = {}
for source in coastIDs :
    stepping_stones_source[source]=[]
    for dest in coastIDs :
        if source != dest :
            path = nx.dijkstra_path(H, source, dest, weight='weight')
            # print(f"{source} --> {dest} : {path}")
            for i in range(1, len(path)-1):
                #if path[i] not in stepping_stones_area[source]:
                stepping_stones_source[source].append(path[i]) 
                if path[i] not in stepping_stones.keys() and path[i] not in coastIDs :
                    stepping_stones[path[i]] = 100/650 #percentage of the total number of short paths
                if path[i] in stepping_stones.keys():
                    stepping_stones[path[i]] += 100/650
```

```python
stepping_stones_dest = {}
for dest in coastIDs :
    stepping_stones_dest[dest]=[]
    for source in coastIDs :
        if source != dest :
            path = nx.dijkstra_path(H, source, dest, weight='weight')
            for i in range(1, len(path)-1):
                stepping_stones_dest[dest].append(path[i]) 
```

#### Printing occurrence of OWFs in shortest paths ####

```python
fig, ax = plt.subplots(figsize=(8, 8))


percent = []
for ID in farmIDs :
    if str(ID) in stepping_stones.keys():
        percent.append(stepping_stones[str(ID)])
    else :
        percent.append(0)
plt.xticks(farmIDs, farmIDs)
plt.bar(farmIDs, percent)
ax.set_ylim([0, 30])

ax.set_title(f'Occurrence in shortest paths between two coastal areas (percentage)')
ax.set_xlabel("Farm ID")
ax.set_ylabel("Occurence (%)")
plt.show()
fig.savefig(f'shortest_path/Shortest_paths_occurence')
```

### Calculating the maximal probability of connection between coastal areas ###
#### With OWFs in the network ####

```python
# calculating the maximum probability of connection between two coastal areas (with the OWF in the network) 
A = np.zeros((nb_c, nb_c))
for i, source in enumerate(coastIDs) :
    for j, dest in enumerate(coastIDs) :
        if source != dest :
            sum_log = nx.dijkstra_path_length(H, source, dest, weight='weight')
            A[i, j] = math.exp(-sum_log) 

xaxis = np.arange(nb_c)
yaxis = np.arange(nb_c) 

fig, ax = plt.subplots(figsize=(8, 8))
A_masked = np.ma.masked_where(A == 0, A) # mask exact zeros
norm = colors.LogNorm(1e-4, 1)
im = ax.matshow(A, cmap='YlOrRd', norm=norm)
im.cmap.set_bad(color='whitesmoke')

# im = ax.matshow(A, cmap='YlOrRd')

plt.colorbar(im, ax=ax, label='Probability of connection')
plt.yticks(yaxis, coastIDs, size='xx-small')
plt.xticks(xaxis, coastIDs, size='xx-small')
ax.set_xlabel("Destination")
ax.set_ylabel("Origin")
ax.set_title(f'Probability of connection between two coastal areas (with OWF)')
plt.savefig("shortest_path/matrix_shortest_path_prob_with")
plt.show()
```

#### Without OWFs in the network ####

```python
# calculating the maximum probability of connection between two coastal areas (without the OWF in the network)
H_without = H.copy()
for ID in farmIDs :
     H_without.remove_node(str(ID))
B = np.zeros((nb_c, nb_c))
for i, source in enumerate(coastIDs) :
    for j, dest in enumerate(coastIDs) :
        if source != dest :
            sum_log = nx.dijkstra_path_length(H_without, source, dest, weight='weight')
            B[i, j] = math.exp(-sum_log) 

xaxis = np.arange(nb_c)
yaxis = np.arange(nb_c) 

fig, ax = plt.subplots(figsize=(8, 8))
B_masked = np.ma.masked_where(B == 0, B) # mask exact zeros
norm = colors.LogNorm(1e-4, 1)
im = ax.matshow(B, cmap='YlOrRd', norm=norm)
im.cmap.set_bad(color='whitesmoke')

# im = ax.matshow(A, cmap='YlOrRd')

plt.colorbar(im, ax=ax, label='Probability of connection')
plt.yticks(yaxis, coastIDs, size='xx-small')
plt.xticks(xaxis, coastIDs, size='xx-small')
ax.set_xlabel("Destination")
ax.set_ylabel("Origin")
ax.set_title(f'Probability of connection between two coastal areas (without OWF)')
plt.savefig("shortest_path/matrix_shortest_path_prob_without")
plt.show()
```

#### Comparison between network with and without OWFs ####

```python
C = (A-B)*100
for i in range(nb_c) :
    for j in range(nb_c) :
        if B[i, j] != 0:
            C[i,j]/B[i, j]

xaxis = np.arange(nb_c)
yaxis = np.arange(nb_c) 

fig, ax = plt.subplots(figsize=(8, 8))
C_masked = np.ma.masked_where(C == 0, C) # mask exact zeros
norm = colors.LogNorm(0.01, 6)
im = ax.matshow(C, cmap='YlOrRd', norm=norm)
im.cmap.set_bad(color='whitesmoke')

# im = ax.matshow(C, cmap='YlOrRd')

plt.colorbar(im, ax=ax, label='Increase of probability of connection (%)')
plt.yticks(yaxis, coastIDs)
plt.xticks(xaxis, coastIDs, rotation=45)
ax.set_xlabel("Destination")
ax.set_ylabel("Origin")
ax.set_title(f'Increase of indirect connectivity when adding OWFs')
plt.savefig("shortest_path/matrix_shortest_path_increase_probability")
plt.show()
```

### Impact of individual farms ###
#### Removing one farm and comparing the results with the whole network (with all OWFs) ####

```python
# calculating the maximum probability of connection between two coastal areas (without one of the OWFs)
for ID in farmIDs :
    H_without = H.copy()
    H_without.remove_node(str(ID))
    B = np.zeros((nb_c, nb_c))
    for i, source in enumerate(coastIDs) :
        for j, dest in enumerate(coastIDs) :
            if source != dest :
                sum_log = nx.dijkstra_path_length(H_without, source, dest, weight='weight')
                B[i, j] = math.exp(-sum_log) 
    
    C = A-B
    xaxis = np.arange(nb_c)
    yaxis = np.arange(nb_c) 
    
    fig, ax = plt.subplots(figsize=(8, 8))
    C_masked = np.ma.masked_where(C == 0, C) # mask exact zeros
    norm = colors.LogNorm(1e-4, 1)
    im = ax.matshow(C, cmap='YlOrRd', norm=norm)
    im.cmap.set_bad(color='whitesmoke')
    
    # im = ax.matshow(C, cmap='YlOrRd')
    
    plt.colorbar(im, ax=ax, label='difference of probability of connection')
    plt.yticks(yaxis, coastIDs, size='xx-small')
    plt.xticks(xaxis, coastIDs, size='xx-small')
    ax.set_xlabel("Destination")
    ax.set_ylabel("Origin")
    ax.set_title(f'Difference between networks with and without OWF N°{ID}')
    plt.savefig(f"shortest_path/difference_individual_farm/matrix_shortest_path_prob_diff_{ID}")
    plt.close()
```

## Other tests/experimentations with the network ##

### Impact of removing farms successively ###

```python
H_without = H.copy()

sorted_farms = dict(sorted(stepping_stones.items(), key=lambda item: item[1]))
prob = {}

sum_prob = 0
cpt = 0
for i, source in enumerate(coastIDs) :
    for j, dest in enumerate(coastIDs) :
        if source != dest :
            sum_log = nx.dijkstra_path_length(H_without, source, dest, weight='weight')                
            sum_prob += math.exp(-sum_log)
            cpt += 1
prob['Initial'] = sum_prob/cpt

# Removing farms that don't appear in shortest paths
for ID in farmIDs:
    if str(ID) not in stepping_stones.keys():
        H_without.remove_node(str(ID))
        
sum_prob = 0
cpt = 0
for i, source in enumerate(coastIDs) :
    for j, dest in enumerate(coastIDs) :
        if source != dest :
            sum_log = nx.dijkstra_path_length(H_without, source, dest, weight='weight')                
            sum_prob += math.exp(-sum_log)
            cpt += 1
prob['Others'] = sum_prob/cpt

# Removing farm from the network sorted by occurence in shortest paths
for ID in sorted_farms.keys():
    H_without.remove_node(ID)
    sum_prob = 0
    cpt = 0
    for i, source in enumerate(coastIDs) :
        for j, dest in enumerate(coastIDs) :
            if source != dest :
                sum_log = nx.dijkstra_path_length(H_without, source, dest, weight='weight')                
                sum_prob += math.exp(-sum_log)
                cpt += 1
    prob[ID] = sum_prob/cpt


fig, ax = plt.subplots(figsize=(10, 10))
plt.plot(prob.keys(), prob.values())

ax.set_title(f'Impact of removing farms successively on the mean probaility of connection between coastal areas')
ax.set_xlabel("Farm ID (sorted by occurrence in shortest paths)")
ax.set_ylabel("Mean probability of connection")
plt.show()
fig.savefig(f'shortest_path/Mean_probability_removing_farms_ascending_cumulated')
```

```python
H_without = H.copy()

prob = {}
sum_prob = 0
cpt = 0
for i, source in enumerate(coastIDs) :
    for j, dest in enumerate(coastIDs) :
        if source != dest :
            sum_log = nx.dijkstra_path_length(H_without, source, dest, weight='weight')                
            sum_prob += math.exp(-sum_log)
            cpt += 1
prob['Initial'] = sum_prob/cpt

# Removing farm from the network sorted by descending occurence in shortest paths
sorted_farms = dict(sorted(stepping_stones.items(), key=lambda item: item[1], reverse=True))

for ID in sorted_farms.keys():
        H_without.remove_node(ID)
        sum_prob = 0
        cpt = 0
        for i, source in enumerate(coastIDs) :
            for j, dest in enumerate(coastIDs) :
                if source != dest :
                    sum_log = nx.dijkstra_path_length(H_without, source, dest, weight='weight')                
                    sum_prob += math.exp(-sum_log)
                    cpt += 1
        prob[ID] = sum_prob/cpt

# Removing farms that don't appear in shortest paths
for ID in farmIDs:
    if str(ID) not in stepping_stones.keys():
        H_without.remove_node(str(ID))
        
sum_prob = 0
cpt = 0
for i, source in enumerate(coastIDs) :
    for j, dest in enumerate(coastIDs) :
        if source != dest :
            sum_log = nx.dijkstra_path_length(H_without, source, dest, weight='weight')                
            sum_prob += math.exp(-sum_log)
            cpt += 1
prob['Others'] = sum_prob/cpt


fig, ax = plt.subplots(figsize=(10, 10))
plt.plot(prob.keys(), prob.values())
# plt.bar(prob.keys(), prob.values())
ax.set_title(f'Impact of removing farms successively on the mean probability of connection between coastal areas')
ax.set_xlabel("Farm ID (sorted by descending occurrence in shortest paths)")
ax.set_ylabel("Mean probability of connection")
plt.show()
fig.savefig(f'shortest_path/Mean_probability_removing_farms_descending_cumulated')
```

### Removing one farm (not cumulated) ###

```python
H_without = H.copy()
sorted_farms = dict(sorted(stepping_stones.items(), key=lambda item: item[1]))
prob = {}
sum_prob = 0
cpt = 0
for i, source in enumerate(coastIDs) :
    for j, dest in enumerate(coastIDs) :
        if source != dest :
            sum_log = nx.dijkstra_path_length(H_without, source, dest, weight='weight')                
            sum_prob += math.exp(-sum_log)
            cpt += 1
prob['Initial'] = sum_prob/cpt

# Removing farms that don't appear in shortest paths
for ID in farmIDs:
    if str(ID) not in stepping_stones.keys():
        H_without.remove_node(str(ID))
        
sum_prob = 0
cpt = 0
for i, source in enumerate(coastIDs) :
    for j, dest in enumerate(coastIDs) :
        if source != dest :
            sum_log = nx.dijkstra_path_length(H_without, source, dest, weight='weight')                
            sum_prob += math.exp(-sum_log)
            cpt += 1
prob['Others'] = sum_prob/cpt

# Removing farm from the network sorted by occurence in shortest paths
for ID in sorted_farms.keys():
    H_without = H.copy()
    H_without.remove_node(ID)
    sum_prob = 0
    cpt = 0
    for i, source in enumerate(coastIDs) :
        for j, dest in enumerate(coastIDs) :
            if source != dest :
                sum_log = nx.dijkstra_path_length(H_without, source, dest, weight='weight')                
                sum_prob += math.exp(-sum_log)
                cpt += 1
    prob[ID] = sum_prob/cpt


fig, ax = plt.subplots(figsize=(10, 10))
# plt.plot(prob.keys(), prob.values(), marker =".", linestyle='None')
plt.bar(prob.keys(), prob.values())
ax.set_ylim([0.0265, 0.0275])
ax.set_title(f'Mean probability of connection between coastal areas after removing one farm')
ax.set_xlabel("Farm ID (sorted by occurrence in shortest paths)")
ax.set_ylabel("Mean probability of connection")
plt.show()
fig.savefig(f'shortest_path/Mean_probability_removing_farms_ascending')
```

### Effects of OWFs depending on the region ###

```python
# Impact of adding farms on regions of coastal areas
regions = {'N_FI':['FI4', 'FI3', 'FI2', 'FI1', 'FI0'],
           'S_FI':['FI8', 'FI7', 'FI6', 'FI5'],           
           'A': ['A1', 'A2', 'A3'],
           'SE_F':['SE10', 'SE11'],
           'S_SE':['SE07', 'SE08', 'SE09', 'SE12', 'SE13', 'SE14'],
           'N_SE': ['SE01', 'SE02', 'SE03', 'SE04', 'SE05', 'SE06']}

  
for region in regions.keys():
    # Searching for stepping stones (farm that appear in shortest paths)
    stepping_stones = {}
    for source in coastIDs :
        for dest in coastIDs :
            if source != dest:
                path = nx.dijkstra_path(H, source, dest, weight='weight')
                for i in range(1, len(path)-1):
                    if path[i] not in stepping_stones.keys() and path[i] not in coastIDs :
                        stepping_stones[path[i]] = 1
                    if path[i] in stepping_stones.keys():
                        stepping_stones[path[i]] += 1
    
    fig, ax = plt.subplots(figsize=(10, 10))
    
    # calculating the initial value of mean indirect connectivity 
    H_without = H.copy()
    for farm in farmIDs:
        H_without.remove_node(str(farm))   
    prob = {}
    sum_prob = 0
    cpt = 0
    for i, source in enumerate(coastIDs) :
        for j, dest in enumerate(coastIDs) :
            if source != dest and (source in regions[region] or dest in regions[region]):
                sum_log = nx.dijkstra_path_length(H_without, source, dest, weight='weight')                
                sum_prob += math.exp(-sum_log)
                cpt += 1
    initial_value = sum_prob/cpt
    
    sorted_farms = dict(sorted(stepping_stones.items(), key=lambda item: item[1]))
    
    # calculating the increase of mean indirect connectivity 
    for ID in sorted_farms:
        H_with = H.copy()
        removed_farms = []
        for farm in farmIDs:
            if str(farm) != ID:
                removed_farms.append(str(farm))
        H_with.remove_nodes_from(removed_farms)
        sum_prob = 0
        cpt = 0
        for i, source in enumerate(coastIDs) :
            for j, dest in enumerate(coastIDs) :
                if source != dest and (source in regions[region] or dest in regions[region]):
                    sum_log = nx.dijkstra_path_length(H_with, source, dest, weight='weight')               
                    sum_prob += math.exp(-sum_log)
                    cpt += 1
        prob[ID] = sum_prob*100/(cpt*initial_value) - 100
    
    plt.bar(prob.keys(), prob.values())
    if region in ['S_FI', 'A', 'S_SE', 'N_FI', 'N_SE']:
        ax.set_ylim([0, 4.5])
    else:
        ax.set_ylim([0, 17])
    
    ax.set_title(f'Increase of mean indirect connectivity after adding one farm (region {region})')
    ax.set_xlabel("Farm added (sorted by occurrence in shortest paths)")
    ax.set_ylabel("Increase of mean indirect connectivity compared to a network without OWF (%)")
    plt.show()
    fig.savefig(f'shortest_path/Region/Mean_probability_adding_farms_region_{region}')
```

```python
# Impact of adding farms on regions of coastal areas
regions = {'N_FI':['FI4', 'FI3', 'FI2', 'FI1', 'FI0'],
           'S_FI':['FI8', 'FI7', 'FI6', 'FI5'],           
           'A': ['A1', 'A2', 'A3'],
           'SE_F':['SE10', 'SE11'],
           'S_SE':['SE07', 'SE08', 'SE09', 'SE12', 'SE13', 'SE14'],
           'N_SE': ['SE01', 'SE02', 'SE03', 'SE04', 'SE05', 'SE06']}

stepping_stones = {}
for source in coastIDs :
    for dest in coastIDs :
        if source != dest:
            path = nx.dijkstra_path(H, source, dest, weight='weight')
            for i in range(1, len(path)-1):
                if path[i] not in stepping_stones.keys() and path[i] not in coastIDs :
                    stepping_stones[path[i]] = 100/650 #percentage of the total number of short paths
                if path[i] in stepping_stones.keys():
                    stepping_stones[path[i]] += 100/650

  
for region in regions.keys():
    
    fig, ax = plt.subplots(figsize=(10, 10)) 
    H_without = H.copy()
    for farm in farmIDs:
        H_without.remove_node(str(farm))
    
    prob = {}
    sum_prob = 0
    cpt = 0
    for i, source in enumerate(coastIDs) :
        for j, dest in enumerate(coastIDs) :
            if source != dest and (source in regions[region] or dest in regions[region]):
                sum_log = nx.dijkstra_path_length(H_without, source, dest, weight='weight')                
                sum_prob += math.exp(-sum_log)
                cpt += 1
    initial_value = sum_prob/cpt
    
    sorted_farms = dict(sorted(stepping_stones.items(), key=lambda item: item[1]))
    
    for ID in sorted_farms:
        H_with = H.copy()
        removed_farms = []
        for farm in farmIDs:
            if str(farm) != ID:
                removed_farms.append(str(farm))
        H_with.remove_nodes_from(removed_farms)
        sum_prob = 0
        cpt = 0
        for i, source in enumerate(coastIDs) :
            for j, dest in enumerate(coastIDs) :
                if source != dest and (source in regions[region] or dest in regions[region]):
                    sum_log = nx.dijkstra_path_length(H_with, source, dest, weight='weight')               
                    sum_prob += math.exp(-sum_log)
                    cpt += 1
        prob[ID] = sum_prob*100/(cpt*initial_value)
    # plt.bar(prob.keys(), prob.values(), label=f'Region:{region}')
    # plt.plot(prob.keys(), prob.values(), label=f'Region:{region}')
    
    # plt.plot(prob.keys(), prob.values(), marker='o', linestyle='None', markersize = 10.0)
    plt.bar(prob.keys(), prob.values())
    ax.set_ylim([100, max(prob.values())+1])
    # ax.legend()
    ax.set_title(f'Increase of mean probability of connection between 2 coastal areas after adding one farm (region {region})')
    # ax.set_title(f'Mean probability of connection between coastal areas after adding one farm')
    ax.set_xlabel("Farm added (sorted by occurrence in shortest paths)")
    ax.set_ylabel("Increase of mean probability of connection compared to a network without OWF (%)")
    plt.show()
    fig.savefig(f'shortest_path/Region/Mean_probability_adding_farms_region_{region}')
    # fig.savefig(f'shortest_path/Mean_probability_adding_farms_region')
```

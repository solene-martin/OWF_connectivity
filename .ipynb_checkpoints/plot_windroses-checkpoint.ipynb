```python
import numpy as np
import pandas as pd
import datetime as dt
import matplotlib as mpl
import matplotlib.pyplot as plt
from windrose import WindroseAxes


dpi = 200
savepic = True
rose_bar_plot = False
force_axis = True
rmax = 20 #12
dr = 5 #2
whole_period = True

bins = [0.0, 6.0, 12., 14., 16., 18.]
##bins = [0, 2.5, 5.0, 7.5, 10., 12.5, 15.]   # for summer months
bins = list(np.arange(0,15,2.5))
nsector = 18
cmap = plt.cm.magma

font = {'family': 'sans-serif',
            'weight': 'normal',
            'size': 6,
    }
mpl.rc('font', **font)

stations = ['MustasaariValassaaret', 'HammarlandMarket', 'KustaviIsokari']
```

```python
datadir = '/projappl/project_2018610/data/AWS/'
outfol = '/projappl/project_2018610/plots/windroses/'
station = stations[2]  #'MustasaariValassaaret'
csvfile = datadir+station+'_19910101-20201231.csv'


df30 = pd.read_csv(csvfile, parse_dates=['utctime'], date_format='%Y%m%dT%H%M%S')

magcol = 'WS_10M_AVG'
dircol = 'WD_10M_AVG'



# rows with mag=0, dir=0 to nan
rows = (df30[magcol]==0) & (df30[dircol]==0)
df30 = df30[~rows]

if whole_period == True:
    # plot whole period
    df = df30
    year1 = 1991
    year2 = 2020
else:
    # take only selected years from dataframe
    year1 = 1991
    year2 = 2000
    T1 = dt.datetime(year1,1,1)
    T2 = dt.datetime(year2+1,1,1)
    rows = (df30.utctime>=T1) & (df30.utctime<T2)
    df = df30[rows]
pngname = station + '_{}-{}'.format(year1,year2)
titledate= '{}-{}'.format(year1,year2)

fig = plt.figure(figsize=(3, 3), dpi=dpi, facecolor='w', edgecolor='w')
rect = [0.1, 0.1, 0.8, 0.8]
ax = WindroseAxes(fig, rect, facecolor='w')
fig.add_axes(ax)
plt.suptitle(u'{}\n{}'.format(station,titledate), y=1.05)
if rose_bar_plot:
    ax.bar(df[dircol],df[magcol],normed=True,opening=0.9,nsector=nsector,bins=bins,cmap=cmap, edgecolor='black', lw=0.2)
else:
    ax.contourf(df[dircol],df[magcol], blowto=False, normed=True, nsector=nsector, bins=bins, cmap=cmap)
    ax.contour(df[dircol],df[magcol], blowto=False, normed=True, nsector=nsector, bins=bins, colors='black', lw=0.5)
if force_axis:
    ticks = range(0,rmax+1,dr)
    ticklabels = [''] + list(map(lambda t: '{}%'.format(t), ticks[1:]))
    ax.set_rmax(rmax)
    ax.set_rticks(ticks)
    ax.set_yticks(ticks)
    ax.set_yticklabels(ticklabels)
# add legend
leg = ax.legend(fontsize=6,title='m s$^{-1}$',loc='upper left', bbox_to_anchor=(-0.07,1.05), framealpha=1.0)
# modify label for first legend entry
legend_texts = leg.get_texts()
newtext = legend_texts[0].get_text().replace('[','(')
legend_texts[0].set_text(newtext)
# save figure
if savepic:
    fig.savefig(outfol+pngname, bbox_inches='tight', dpi=dpi)
    plt.close(fig)
```

```python
####Script to plot the windrose of each year for one station

savepic = True

years2plot = list(range(1990,2021))
years2plot = list(range(2010,2025))
#years2plot = [2012,2013,2017,2018]

# defining different seasons
seasons = {
    'JFM': [1,2,3],
    'AMJ': [4,5,6],
    'JAS': [7,8,9],
    'OND': [10,11,12],
    'DJF': [12,1,2],
    'MAM': [3,4,5],
    'JJA': [6,7,8],
    'SON': [9,10,11],
    'JJAS': [6,7,8,9]
}
# here using winter: Dec-Feb, spring: Mar-May, summer: Jun-Aug, autumn: Sep-Nov
seasons2plot = ['DJF','MAM','JJA','SON']


# set station
for station in stations:
    
    print('plotting windroses for '+station)
    
    # set paths
    datadir = '/projappl/project_2018610/data/AWS/'
    outfol = '/projappl/project_2018610/plots/windroses/'+ station +'/'
    csvfile = datadir+station+'_19910101-20241231.csv'
    
    # read station file
    df_all = pd.read_csv(csvfile, parse_dates=['utctime'], date_format='%Y%m%dT%H%M%S')
    
    magcol = 'WS_10M_AVG'
    dircol = 'WD_10M_AVG'
    
    # rows with mag=0, dir=0 to nan
    rows = (df_all[magcol]==0) & (df_all[dircol]==0)
    df_all = df_all[~rows]

    for year1 in years2plot:
        # take given year from dataframe
        year2 = year1+1
        T1 = dt.datetime(year1,1,1)
        T2 = dt.datetime(year2+1,1,1)
        rows = (df_all.utctime>=T1) & (df_all.utctime<T2)
        df = df_all[rows]
    
        # ------------
        # yearly roses
        # ------------
        pngname = station + '_{}'.format(year1)
        titledate= '{}'.format(year1)
        print(outfol+pngname)
        
        fig = plt.figure(figsize=(3, 3), dpi=dpi, facecolor='w', edgecolor='w')
        rect = [0.1, 0.1, 0.8, 0.8]
        ax = WindroseAxes(fig, rect, facecolor='w')
        fig.add_axes(ax)
        plt.suptitle(u'{} {}'.format(station,titledate), y=1.025)
        if rose_bar_plot:
            ax.bar(df[dircol],df[magcol],normed=True,opening=0.9,nsector=nsector,bins=bins,cmap=cmap, edgecolor='black', lw=0.2)
        else:
            ax.contourf(df[dircol],df[magcol], blowto=False, normed=True, nsector=nsector, bins=bins, cmap=cmap)
            ax.contour(df[dircol],df[magcol], blowto=False, normed=True, nsector=nsector, bins=bins, colors='black', lw=0.5)
        if force_axis:
            ticks = range(0,rmax+1,dr)
            ticklabels = [''] + list(map(lambda t: '{}%'.format(t), ticks[1:]))
            ax.set_rmax(rmax)
            ax.set_rticks(ticks)
            ax.set_yticks(ticks)
            ax.set_yticklabels(ticklabels)
        # add legend
        leg = ax.legend(fontsize=6,title='m s$^{-1}$',loc='upper left', bbox_to_anchor=(-0.07,1.05), framealpha=1.0)
        # modify label for first legend entry
        legend_texts = leg.get_texts()
        newtext = legend_texts[0].get_text().replace('[','(')
        legend_texts[0].set_text(newtext)
        # save figure
        if savepic:
            fig.savefig(outfol+pngname, bbox_inches='tight', dpi=dpi)
            plt.close(fig)
    
        # --------------
        # seasonal roses
        # --------------
        pngname = station + '_seasonal_{}'.format(year1)
        titledate = '{}'.format(year1)
        print(outfol+pngname)
        
        fig,axes = plt.subplots(2,2,figsize=(6,6), dpi=dpi, facecolor='w', edgecolor='w')
        plt.subplots_adjust(hspace=0.3,wspace=0.4)
        plt.suptitle(u'{} {}'.format(station,titledate))#y=1.05)
        for season,ax in zip(seasons2plot,axes.flatten()):
            rect = ax.get_position()
            ax.axis('off')
            wax = WindroseAxes(fig, rect, facecolor='w')
            fig.add_axes(wax)
            wax.set_title(season,y=1.05,backgroundcolor='white')
    
            months = seasons[season]
            #print(months)
            rows = df.utctime.dt.month.isin(months)
            
            if rose_bar_plot:
                wax.bar(df[rows][dircol],df[rows][magcol],normed=True,opening=0.9,nsector=nsector,bins=bins,cmap=cmap, edgecolor='black', lw=0.2)
            else:
                wax.contourf(df[rows][dircol],df[rows][magcol], blowto=False, normed=True, nsector=nsector, bins=bins, cmap=cmap)
                wax.contour(df[rows][dircol],df[rows][magcol], blowto=False, normed=True, nsector=nsector, bins=bins, colors='black', lw=0.5)
            if force_axis:
                ticks = range(0,rmax+1,dr)
                ticklabels = [''] + list(map(lambda t: '{}%'.format(t), ticks[1:]))
                wax.set_rmax(rmax)
                wax.set_rticks(ticks)
                wax.set_yticks(ticks)
                wax.set_yticklabels(ticklabels)
            # add legend
            if False:
                leg = wax.legend(fontsize=6,title='m s$^{-1}$',loc='upper left', bbox_to_anchor=(-0.07,1.05), framealpha=1.0)
                # modify label for first legend entry
                legend_texts = leg.get_texts()
                newtext = legend_texts[0].get_text().replace('[','(')
                legend_texts[0].set_text(newtext)
    
        # save figure
        if savepic:
            fig.savefig(outfol+pngname, bbox_inches='tight', dpi=dpi)
            plt.close(fig)
    
        # ------------------------------------
        # roses for June-July-August-September
        # ------------------------------------
        season = 'JJAS'
        pngname = station + '_{}_{}'.format(season,year1)
        titledate= '{} {}'.format(season,year1)
        print(outfol+pngname)
    
        months = seasons[season]
        rows = df.utctime.dt.month.isin(months)
        
        fig = plt.figure(figsize=(3, 3), dpi=dpi, facecolor='w', edgecolor='w')
        rect = [0.1, 0.1, 0.8, 0.8]
        ax = WindroseAxes(fig, rect, facecolor='w')
        fig.add_axes(ax)
        plt.suptitle(u'{} {}'.format(station,titledate),y=1.025)
        if rose_bar_plot:
            ax.bar(df[rows][dircol],df[rows][magcol],normed=True,opening=0.9,nsector=nsector,bins=bins,cmap=cmap, edgecolor='black', lw=0.2)
        else:
            ax.contourf(df[rows][dircol],df[rows][magcol], blowto=False, normed=True, nsector=nsector, bins=bins, cmap=cmap)
            ax.contour(df[rows][dircol],df[rows][magcol], blowto=False, normed=True, nsector=nsector, bins=bins, colors='black', lw=0.5)
        if force_axis:
            ticks = range(0,rmax+1,dr)
            ticklabels = [''] + list(map(lambda t: '{}%'.format(t), ticks[1:]))
            ax.set_rmax(rmax)
            ax.set_rticks(ticks)
            ax.set_yticks(ticks)
            ax.set_yticklabels(ticklabels)
        # add legend
        leg = ax.legend(fontsize=6,title='m s$^{-1}$',loc='upper left', bbox_to_anchor=(-0.07,1.05), framealpha=1.0)
        # modify label for first legend entry
        legend_texts = leg.get_texts()
        newtext = legend_texts[0].get_text().replace('[','(')
        legend_texts[0].set_text(newtext)
        # save figure
        if savepic:
            fig.savefig(outfol+pngname, bbox_inches='tight', dpi=dpi)
            plt.close(fig)
```
